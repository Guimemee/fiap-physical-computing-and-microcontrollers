> **Enunciado Original:**
> Desenvolver aplicação multitarefa preemptiva no FreeRTOS com xTaskCreatePinnedToCore() distribuindo tarefas de controle crítico no Core 0 e conectividade Wi-Fi no Core 1 com filas de mensagens (Queues) e semáforos binários.

### Formulação Matemática Formal
$$\text{Core 0 (Controle)}: \text{TaskPID (Prioridade 5)}, \quad \text{Core 1 (Rede)}: \text{TaskMQTT (Prioridade 2)}$$


### Passo 1: Fixação de Tarefas por Afinidade de Núcleo (Core Pinning)
xTaskCreatePinnedToCore() isola a malha de controle PID no Core 0 livre de interrupções de pilha Wi-Fi.


### Passo 2: Filas de Mensagens (xQueueSend / xQueueReceive)
Buffer FIFO protegido entre tarefas prevenindo acessos concorrentes a dados de telemetria.


### Passo 3: Notificações Diretas entre Tarefas (Task Notifications)
Sincronização ultrarrápida com xTaskNotifyGive() com sobrecarga 45% menor que semáforos.


### Passo 4: Monitoramento de Stack High Water Mark
uxTaskGetStackHighWaterMark() para assegurar que nenhuma tarefa sofra estouro de pilha (Stack Overflow).


### Passo 5: Conclusão de Multitarefa
Estabilidade perfeita com latência de resposta a interrupções (ISR latency) inferior a 2,5 \mu\text{s}.


```python
import numpy as np
import matplotlib.pyplot as plt

tarefas = ['TaskPID (Core 0)', 'TaskSensores (Core 0)', 'TaskMQTT (Core 1)', 'TaskDisplay (Core 1)']
prioridades = [5, 4, 2, 1]
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#130d29')
ax.set_facecolor('#130d29')
bars = ax.barh(tarefas, prioridades, color='#8b5cf6', height=0.45)
ax.set_title('Hierarquia de Prioridades Preemptivas no FreeRTOS', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Nível de Prioridade (Maior = Mais Urgente)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='x', color='#352466', alpha=0.5)
for bar in bars:
    wval = bar.get_width()
    ax.text(wval + 0.1, bar.get_y() + bar.get_height()/2.0, f'Prioridade {wval}', ha='left', va='center', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
