> **Enunciado Original:**
> Dimensionar o periférico LEDC PWM do ESP32 com resolução de 14 bits na frequência de 5 kHz para controle de velocidade e posicionamento angular suave de atuadores robóticos.

### Formulação Matemática Formal
$$f_{PWM} = \frac{f_{clock}}{Prescaler \times 2^{\text{Resolution}}}, \quad 5000 = \frac{80.000.000}{1 \times 2^{14}} = \frac{80.000.000}{16384} \approx 4882,8 \text{ Hz}$$


### Passo 1: Clock Base do Periférico (APB Clock 80 MHz)
Seleção de fonte de clock estável independente de flutuações de CPU.


### Passo 2: Definição de Resolução de 14 Bits (16.384 Níveis)
Permite controle fino de Duty Cycle de 0 a 100% com incremento de apenas 0,0061%.


### Passo 3: Geração de Fade Hardware Automático
Uso da função ledc_set_fade_with_time() para transições angulares suaves sem consumir processamento do núcleo.


### Passo 4: Proteção contra Emissão Eletromagnética
Frequência calibrada em 5 kHz eliminando ruídos audíveis em bobinas e atuadores.


### Passo 5: Conclusão Mecatrônica
Movimentação do braço robótico com resolução micrométrica sem vibrações residuais.


```python
import numpy as np
import matplotlib.pyplot as plt

duty_pct = np.linspace(0, 100, 100)
posicao_graus = (duty_pct / 100) * 180
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#130d29')
ax.set_facecolor('#130d29')
ax.plot(duty_pct, posicao_graus, color='#8b5cf6', lw=2.5, label='Posição Angular (Graus)')
ax.set_title('Linearidade do Posicionamento Angular via PWM 14-Bit', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Duty Cycle Aplicado (%)', color='#94a3b8')
ax.set_ylabel('Posição da Junta (0° a 180°)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#352466', alpha=0.5)
ax.legend(facecolor='#211642', edgecolor='#8b5cf6', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
