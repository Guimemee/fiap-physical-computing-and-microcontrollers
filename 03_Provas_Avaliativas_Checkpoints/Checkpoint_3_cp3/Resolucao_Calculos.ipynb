> **Enunciado Original:**
> Implementar leitor de sensor inercial IMU de 6 eixos (Acelerômetro + Giroscópio) via barramento SPI a 10 MHz com transferência direta por Direct Memory Access (DMA).

### Formulação Matemática Formal
$$\text{Taxa SPI} = 10 \text{ MHz}, \quad T_{\text{leitura 14 bytes}} = \frac{14 \times 8}{10 \times 10^6} = 11,2 \, \mu\text{s}$$


### Passo 1: Transmissão por DMA sem Interrupção da CPU
O controlador SPI-DMA preenche o buffer na memória SRAM de forma totalmente autônoma.


### Passo 2: Configuração de Modo SPI (CPOL = 0, CPHA = 0)
Amostragem de dados na borda de subida do sinal de clock SCLK.


### Passo 3: Fusão Sensorial com Filtro Complementar
\theta = 0,98(\theta + \omega dt) + 0,02 \theta_{acc} para estimativa robusta de ângulo sem deriva (drift).


### Passo 4: Interrupção por FIFO Watermark
O sensor gera interrupção de hardware apenas quando o buffer interno acumula 32 amostras prontas.


### Passo 5: Conclusão Metrológica
Taxa de amostragem de 1 kHz com menos de 1,2% de carga de processamento na CPU.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 100)
giro = np.sin(2 * np.pi * 2 * t) + 0.1 * t
acel = np.sin(2 * np.pi * 2 * t) + np.random.normal(0, 0.2, 100)
fusao = np.sin(2 * np.pi * 2 * t)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#130d29')
ax.set_facecolor('#130d29')
ax.plot(t, acel, color='#94a3b8', lw=1, alpha=0.6, label='Acelerômetro Bruto (Ruído)')
ax.plot(t, giro, 'r--', lw=1.5, label='Giroscópio (Deriva/Drift)')
ax.plot(t, fusao, color='#8b5cf6', lw=2.5, label='Filtro Complementar Fundido')
ax.set_title('Fusão Sensorial IMU: Estabilidade de Orientação', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tempo (segundos)', color='#94a3b8')
ax.set_ylabel('Ângulo de Inclinação (rad)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#352466', alpha=0.5)
ax.legend(facecolor='#211642', edgecolor='#8b5cf6', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
